**Univariate Analysis → distribution of single variables (countplot, histogram, boxplot)**

```python
# Countplot
plt.figure(figsize=(5,3))
sns.countplot(x='product_category', data=df,color='violet')
plt.xticks(rotation=45)
plt.title('Product Category Distribution')
plt.show()
```

**Interpretation of Count plot:**

Countplot shows the frequency distribution of categorical data to identify the most common category.
*   The chart uses the Product Category feature from the dataset.
*   The Countplot shows the number of orders placed in each product category.It helps measure the frequency of sales across different categories.
*   This helps identify popular and less popular product categories.
*   It helps businesses focus on high-demand products,improve inventory planning and create better marketing strategies.


```python
# Histogram
sns.histplot(df['revenue'], bins=20, kde=True,color='green')
plt.title('Distribution of Revenue')
plt.xlabel('Revenue')
plt.ylabel('Frequency')
plt.show()
```

**Interpretation of Histogram:**

Histogram shows how revenue values are distributed and helps understand the data spread.

*   The Chart uses the revenue feature from the dataset.
*   The histogram shows the distribution of revenue values across all e-commerce transactions.
*   It shows where most sales revenue falls and it indicates high-value sales as well.
*   From this chart, we can identify the common revenue range, understand sales distibution patterns, and detect whether the business depends on small or large transactions.

```python
# Boxplot
sns.boxplot(x='product_category', y='revenue', data=df,color='orange')
plt.xticks(rotation=45)
plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Revenue')
plt.show()
```

**Interpretation of Boxplot:**

Boxplot shows the spread, median, and outliers of revenue across product categories.
*   The chart uses Product category on the x-axis and Revenue on the y-axis.
*   The box plot shows the distibution of revenue across different product categories. It displays the median revenue,spread of revenue values, and outliers in each category.
*   The chart indicates that revenue distibution is fairly similar across product categories.
*   From this chart, we can compare revenue consistency between categories, identify categories with stable or variable revenue, and detect unusual transactions.

**Bivariate Analysis → relation between two variables (scatterplot, barplot, correlation heatmap)**

```python
# Scatter plot
sns.scatterplot(x='quantity', y='revenue', data=df,color='brown')
plt.xlabel('Quantity')
plt.ylabel('Revenue')
plt.title('Comparison between quantity and revenue')
plt.grid(True)
plt.show
```

**Interpretation of Scatter plot:**

Scatter plot shows the relationship between quantity sold and revenue generated.
*   The chart uses quantity (x-axis) and revenue (y-axis).
*   The scatter plot shows the relationship between the number of products sold and the revenue generated from those sales.
*   The chart indicates a positive relationship between quantity and revenue. As the quantity increases, revenue also tends to increase.
*   From this chart, we can understand how product quantity affects revenue generation.

```python
# Bar plot
plt.figure(figsize=(5,3))
sns.barplot(x='region', y='discount', data=df,palette='pastel')
plt.xticks(rotation=45)
plt.title('Region wise Discount')
plt.xlabel('Region')
plt.ylabel('Discount')
plt.show()
```

**Interpretation of Bar plot:**

Barplot compares average values like discount or revenue across different categories or regions.
*   This bar chart shows the average discount given in each region. The x-axis represents different regions, and the y-axis represents the discount value.
*   The chart shows that discounts are almost similar across all regions, with only small differences.
*   The chart uses region (x-axis) and discount (y-axis).
*   This chart helps us understand how discounts are distributed across regions. It shows the company follows a nearly equal discount strategy in all regions.

```python
# Correlation heatmap
plt.figure(figsize=(6,5))
corr = df[['revenue','quantity','discount']].corr()
sns.heatmap(corr, annot=True,cmap='coolwarm')
plt.title('Correlation between numerical features')
plt.show()
```

**Interpretation of Correlation heatmap:**

Correlation heatmap shows the strength and direction of relationship between numeical features.
*   This heatmap shows the correlation between three numerical features:Revenue,Quantity,and Discount.Correlation range from -1 to 1.
* What is it saying?
   1.  Revenue and quantity(0.6): Moderate positive correlation
   2.  Revenue and quantity(0.11): Very weak negative correlation.
   3.  Revenue and quantity(-0.0043):Almost no relationship.
*   The features used in this map Revenue, Quantity and Discount.
*   This heatmap helps identify how important each feature is related to another. It shows that quantiity is the main driver of revenuee, while discounts has very little impact on both revenue and quantity.


**Multivariate Analysis → relation among 3+ variables (pairplot, grouped analysis, pivot tables, advanced plots)**


```python
# Pair plot
sns.pairplot(df[['revenue','quantity','discount','region']],hue='region')
plt.show()
```

**Interpretation of Pairplot:**

Pairplot shows pairwise relationships and distributions among multiple numerical variables.
*   A pairplot shows the relationship between multiple numerical variables at the same time.
*   A pairplot shows that Revenue and Quantity have a positive relationship, meaning when quantity increases, revenue also increases.
*   Features used in this pairplot Revenue, Quantity, Discount and Region.
*   This chart helps identify patterns among multiple variables together. It shows sales performance, product quantity impact and discount influence in one view.

```python
# Grouped analysis
grouped_data = df.groupby(['product_category','region'])['revenue'].sum().reset_index()
sns.barplot(x='product_category',y='revenue',hue='region',data=grouped_data)
plt.xticks(rotation=45)
plt.title('Total Revenue by Product Category and Region')
plt.show()
```

**Interpretation of Grouped analysis:**

Grouped analysis compares revenue across multiple categories and regions together.
*   This grouped bar chart compares total revenue across different product categories for each region(East, North, South, West).Each colour represents region.
*   The chart shows that some product categories generate higher revenue than others, and revenue peformance varies by region.
*   Features used in this analysis Product category, Region, and Revenue.
*   This chart helps identify which product categories perform better in each regionn and where the business earns more revenue.

```python
# Pivot Table analysis
pivot_table = pd.pivot_table(df, values='revenue', index='product_category', columns='region', aggfunc='sum')
print(pivot_table)
```

**Interpretation of Pivot table analysis:**

Pivot table analysis summarizes revenue values based on product category and region for easy comparison
*   This pivot table summarizes total revenue for each product category across different regions(East, North, South and West).It helps compare sales performance in a astructured format.
*   The Table shows that Electronics has the highest revenue in all regions, especially in the South and North.
*   Features used in this table Product category, Region and Revenue.
*   This analysis shows which product category performs best in each region and where sales and strongest.

```python
# Adavanced plot(Heat map from pivot table)
sns.heatmap(pivot_table, annot=True, fmt='.2f')
plt.title('Total Revenue by Product Category and Region')
plt.show()
```

**Interpretation of Advanced plot(Heatmap from pivot table):**

Heatmap from pivot table visualizes summarized revenue data to quickly identify high and low performing areas.
*   The heatmap shows the total revenue generated by each product category across different regions.
*   The chart shows how revenue changes based on both category and region. It helps identify which product performs best in which region.
*   Features used Product category, Region and Revenue.
*   It shows best performing product category and region, Revenue comparison between categories and regions, Business areas where sales are strongest and weakest.
